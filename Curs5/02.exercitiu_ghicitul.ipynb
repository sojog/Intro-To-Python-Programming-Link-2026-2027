# Rezolvarea exercitiului

# Refactorizare 1

```python
nr_calculator = int(random.random() * 10)
nr_utilizator = input("Introduceti un numar")

while not nr_utilizator.isnumeric():
    nr_utilizator = input("Introduceti un numar")

nr_utilizator = int(nr_utilizator)

print("Ai introdus:", nr_utilizator, "iar calculator a ales:", nr_calculator)

nr_utilizator == nr_calculator
```

# Refactorizare 2 - numarul introdus trebuie sa fie intre 1 si 10

```python
nr_calculator = int(random.random() * 10)
nr_utilizator = input("Introduceti un numar intre 1 si 10")

# True or ...

while not nr_utilizator.isnumeric() or not (int(nr_utilizator) >= 1 and int(nr_utilizator) <= 10):
    nr_utilizator = input("Introduceti un numar intre 1 si 10")

nr_utilizator = int(nr_utilizator)

print("Ai introdus:", nr_utilizator, "iar calculator a ales:", nr_calculator)

nr_utilizator == nr_calculator
```
