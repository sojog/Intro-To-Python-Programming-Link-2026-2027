# Varianta 1

### Generarea unui număr random între 0 și 1 se face cu funcția random din modulul random

```python
int(random.random() * 10)
```

```python
nr_calculator = int(random.random() * 10) + 1
nr_utilizator = input("Introduceti un numar intre 1 si 10")

# True or ...

a_ghicit = False


while not a_ghicit:

    while not nr_utilizator.isnumeric() or not (int(nr_utilizator) >= 1 and int(nr_utilizator) <= 10):
        nr_utilizator = input("Introduceti un numar intre 1 si 10")

    nr_utilizator = int(nr_utilizator)
    
    if nr_utilizator == nr_calculator:
        print("Ai introdus:", nr_utilizator, "iar calculator a ales:", nr_calculator)
        a_ghicit = True
        
    else:
        print("Ai introdus:", nr_utilizator, " nu este varianta corecta")
        nr_utilizator = input("Introduceti un numar intre 1 si 10")
    

```

```python
nr_calculator = random.randint(1, 10)
nr_utilizator = input("Introduceti un numar intre 1 si 10")

# True or ...


while True:

    while not nr_utilizator.isnumeric() or not (int(nr_utilizator) >= 1 and int(nr_utilizator) <= 10):
        nr_utilizator = input("Introduceti un numar intre 1 si 10")

    nr_utilizator = int(nr_utilizator)
    
    if nr_utilizator == nr_calculator:
        print("Ai introdus:", nr_utilizator, "iar calculator a ales:", nr_calculator)
        break
        
    else:
        print("Ai introdus:", nr_utilizator, " nu este varianta corecta")
        nr_utilizator = input("Introduceti un numar intre 1 si 10")
```
