# Range-ul reprezinta o colectie de numere intregi

```python
for i in range(0, 3):
    print("i = ", i, type(i))
```

# Stringul este tot o colectie

```python
for i in "romania":
    print("i = ", i, type(i))
```

# Lista și tuple-ul sunt tot colecții

```python
for i in (20, 33, 1, -100, 45): # tuple
    print("i = ", i, type(i))
```

```python
for i in [20, 33, 1, -100, 45]: # lista
    print("i = ", i, type(i))
```

# For functioneaza cu break si continue

```python
for i in range(50):
    print("i = ", i)
    if i == 3:
        break
```

```python
for i in "romania":
    if i == "a":
        continue
    print("i = ", i, type(i))
```
