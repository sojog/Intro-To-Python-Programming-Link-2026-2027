# Varianta in care liste sunt de lungimi diferite

```python
lista_a = [1, 2, 3]
lista_b = [4, 5, 6, 7, 8, 9]

lista_sume = []

for i in range(len(lista_b)):
    lista_sume.append(lista_a[i] + lista_b[i])

lista_sume
```

```python
##########
lista_a = [1, 2, 3] # input
lista_b = [4, 5, 6, 3, 3, 3, 8, 8, 9]
##########

if len(lista_a) < len(lista_b):
    print("Prima conditie este adevarata")
    lista_min = lista_a
    lista_max = lista_b
else:
    print("A doua conditie este adevarata")
    lista_min = lista_b
    lista_max = lista_a



lista_sume = []

for i in range(len(lista_min)):
    lista_sume.append(lista_a[i] + lista_b[i])


for i in range(len(lista_min), len(lista_max)):
    lista_sume.append(lista_max[i])

lista_sume

```
