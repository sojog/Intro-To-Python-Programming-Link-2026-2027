# Varianta clasica

```python
numere = [3, 7, 1, 9, 2, 4, 5, 12]
impare = []
pare = []
```

```python
for nr in numere:
    if nr % 2 == 0:
        pare.append(nr)
    else:
        impare.append(nr)

```

```python
print("pare:", pare)
print("impare:", impare)
```

# Varianta Pythonica (LIST COMPREHENTION)

```python
numere = [3, 7, 1, 9, 2, 4, 5, 12]
impare = [ nr for nr in numere  if nr % 2 == 1 ]
pare = [ nr for nr in numere if nr % 2 == 0 ]

print("pare:", pare)
print("impare:", impare)
```
