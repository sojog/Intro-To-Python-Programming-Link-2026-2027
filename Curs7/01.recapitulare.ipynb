# Colectiile mutabile (lista) - au functii de sort in place

```python
i = 1
while i < 6:
    i += 1
    if i == 3 :
        break
else:
    i+=1
    
i += 1
print ("Value of i is : ", i)
```

```python
i = 1
while i < 6:
    i += 1
    if i == 3 :
        continue
else: # asta se executa dupa ce conditia nu mai este adevarat, dar nu se executa cand se face break
    i+=1
    
i += 1
print ("Value of i is : ", i)
```
