```python
o_lista = ["alb", "verde", "galben", "negru"]

for i in enumerate(o_lista):
    print(i)
```

```python
o_lista = ["alb", "verde", "galben", "negru"]

for cheie, valoare in enumerate(o_lista):
    print(cheie, valoare)
```

```python
un_dictionar = { "perete" : "alb", "patrunjel" :  "verde",  "ou" : "galben", "telefon" : "negru"}

for cheie, valoare in enumerate(un_dictionar):
    print("cheia:", cheie, "valoare:", valoare)
```
